```python


def generate(x):
    response: ChatResponse = chat(model='deepseek-r1:8b', messages=[
      {
        'role': 'user',
        'content': x,
      },
    ])
    return response['message']['content']

```

```python
df = pd.read_csv("Data/reference_mathematical-olympiad-progress-prize.csv")
ans_all = []
for i in range(len(df["problem"])):
    print("Doing %d of %d"%(i,len(df["problem"])))
    ans_i = generate(x=df["problem"][0])
    ans_all.append(ans_i)
    print(ans_i)
```
